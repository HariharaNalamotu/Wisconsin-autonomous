```python
import numpy as np
import matplotlib.pyplot as plt

def plot_birds_eye(points):
    """
    points: list of numpy arrays
            each array has 4 values [x, y, z, _]
    """

    arr = np.array(points)

    if arr.ndim != 2 or arr.shape[1] < 2:
        raise ValueError("Each point must have at least x and y values.")

    # Keep only rows where x and y are finite
    valid_mask = np.isfinite(arr[:, :2]).all(axis=1)
    valid_points = arr[valid_mask][:, :2]

    if valid_points.shape[0] == 0:
        print("No valid points to plot.")
        return

    x = valid_points[:, 0]
    y = valid_points[:, 1]

    plt.figure()
    plt.plot(x, y)
    plt.title("Bird's Eye View (Top-Down)")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.axis("equal")  # keeps aspect ratio correct
    plt.grid(True)

    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def ego_trajectory_from_light_points(light_points_cam):
    """
    light_points_cam: list/array of [X,Y,Z] points (traffic light) in camera frame.
                      Camera frame: +X forward, +Y right, +Z up.
    Returns: Nx2 array of ego positions [x,y] in world ground frame (X forward, Y left).
    """
    P = np.asarray(light_points_cam)  # (N,3) or (N,>=3)
    print(len(P))

    # keep only X,Y and make sure finite
    XY = P[:, :2]
    print(len(XY))
    valid = np.isfinite(XY).all(axis=1)
    #XY = XY[valid]
    print(len(XY))

    if XY.shape[0] < 2:
        raise ValueError("Not enough valid points to compute trajectory.")

    # Convert camera ground vector (X forward, Y right) -> (X forward, Y left)
    # v points from car to light in the ground plane
    v = np.column_stack([XY[:, 0], -XY[:, 1]])  # [X, -Y]
    print(len(v))

    # Compute rotation so that initial car->light vector aligns with +X at t=0
    v0 = v[0]
    print(len(v))
    theta0 = np.arctan2(v0[1], v0[0])  # angle of v0 in (X, Y_left)

    c, s = np.cos(-theta0), np.sin(-theta0)
    R = np.array([[c, -s],
                  [s,  c]])

    v_world = (R @ v.T).T  # rotate every vector

    # Light is at world origin, so car position is negative of (car->light) vector
    car_xy = -v_world
    return car_xy

def plot_birds_eye_xy(xy, title="Ego Trajectory (BEV)"):
    x = xy[:, 0]
    y = xy[:, 1]
    plt.figure()
    plt.plot(x, y, marker='.', linewidth=1)
    plt.title(title)
    plt.xlabel("X (forward)")
    plt.ylabel("Y (left)")
    plt.axis("equal")
    plt.grid(True)
    plt.show()

# --- use your existing ego_centric list ---
# ego_centric currently stores xyz[v,u] each frame
ego_xy = ego_trajectory_from_light_points(ego_centric)
plot_birds_eye_xy(ego_xy)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# ego_xy is Nx2 array of ego positions in BEV with traffic light at origin
# If you haven't computed it yet, do:
# ego_xy = ego_trajectory_from_light_points(ego_centric)

def animate_trajectory_30fps(xy, out_mp4="trajectory.mp4", fps=30, show=True):
    xy = np.asarray(xy)
    valid = np.isfinite(xy).all(axis=1)
    xy = xy[valid]

    if xy.shape[0] < 2:
        raise ValueError("Not enough valid points to animate.")

    # --- If your data isn't exactly 30Hz, resample to exactly 30 points/sec ---
    # Assume original samples are one per frame with constant dt.
    # Create a resampled sequence that has exactly one point per 1/fps seconds.
    t_orig = np.arange(len(xy), dtype=float)          # frame index timebase
    t_new = np.linspace(0, len(xy) - 1, len(xy))      # default: same count
    # If you *want* strict 30 Hz for a 10 sec clip, do this instead:
    # t_new = np.linspace(0, len(xy) - 1, int(10 * fps), endpoint=False)

    x_new = np.interp(t_new, t_orig, xy[:, 0])
    y_new = np.interp(t_new, t_orig, xy[:, 1])
    xy = np.column_stack([x_new, y_new])

    # --- setup plot ---
    fig, ax = plt.subplots()
    ax.set_title("Ego Trajectory in BEV (Traffic Light at Origin)")
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.grid(True)
    ax.axis("equal")

    # plot the origin (traffic light ground point)
    ax.scatter([0], [0], s=60, marker="*", label="Traffic Light (Origin)")

    # dynamic artists: trail + current point
    (trail_line,) = ax.plot([], [], linewidth=2, label="Ego Path")
    (cur_point,)  = ax.plot([], [], marker="o", markersize=6, linestyle="None", label="Ego Vehicle")

    ax.legend(loc="best")

    # set reasonable bounds with padding
    xmin, ymin = xy.min(axis=0)
    xmax, ymax = xy.max(axis=0)
    pad_x = 0.05 * (xmax - xmin + 1e-9)
    pad_y = 0.05 * (ymax - ymin + 1e-9)
    ax.set_xlim(xmin - pad_x, xmax + pad_x)
    ax.set_ylim(ymin - pad_y, ymax + pad_y)

    def init():
        trail_line.set_data([], [])
        cur_point.set_data([], [])
        return trail_line, cur_point

    def update(i):
        # trail up to i
        trail_line.set_data(xy[:i+1, 0], xy[:i+1, 1])
        # current point
        cur_point.set_data([xy[i, 0]], [xy[i, 1]])
        return trail_line, cur_point

    interval_ms = 1000 / fps
    anim = FuncAnimation(
        fig, update, frames=len(xy),
        init_func=init, blit=True, interval=interval_ms
    )

    # --- save ---
    try:
        from matplotlib.animation import FFMpegWriter
        writer = FFMpegWriter(fps=fps, bitrate=1800)
        anim.save(out_mp4, writer=writer)
        print(f"Saved: {out_mp4}")
    except Exception as e:
        print("Could not save MP4 with ffmpeg. Error:", e)
        print("Tip: install ffmpeg, or save as GIF using PillowWriter.")

    if show:
        plt.show()
    else:
        plt.close(fig)

# Example usage:
# animate_trajectory_30fps(ego_xy, out_mp4="trajectory.mp4", fps=30, show=True)
```

```python
import numpy as np

def robust_sigma(x):
    """Robust std estimate using MAD (works well with outliers)."""
    med = np.median(x)
    mad = np.median(np.abs(x - med)) + 1e-9
    return 1.4826 * mad  # MAD -> sigma for Gaussian-ish data

def estimate_gate_from_first_frame(xyz0, uv0, roi_half=25, zmin=-2.0, zmax=3.0):
    """
    xyz0: (H,W,3) xyz in meters
    uv0: (u,v) pixel location of cart in frame 0
    roi_half: half-size of ROI square in pixels around uv0
    Returns:
      center0: robust 3D center [X,Y,Z]
      (dx,dy,dz): gate half-widths in meters (auto)
    """
    u0, v0 = uv0
    H, W, _ = xyz0.shape

    u1 = int(np.clip(u0 - roi_half, 0, W-1))
    u2 = int(np.clip(u0 + roi_half, 0, W-1))
    v1 = int(np.clip(v0 - roi_half, 0, H-1))
    v2 = int(np.clip(v0 + roi_half, 0, H-1))

    pts = xyz0[v1:v2+1, u1:u2+1].reshape(-1, 3)
    pts = pts[np.isfinite(pts).all(axis=1)]

    # optional: remove extreme Z (ground/sky) if your Z is reasonable
    pts = pts[(pts[:,2] > zmin) & (pts[:,2] < zmax)]
    if len(pts) < 30:
        raise ValueError("Too few valid points in ROI to estimate gate. Increase roi_half.")

    center0 = np.median(pts, axis=0)

    # measure spread around center
    d = pts - center0
    sx, sy, sz = robust_sigma(d[:,0]), robust_sigma(d[:,1]), robust_sigma(d[:,2])

    # gate = k * sigma + a safety margin
    k = 4.0
    margin = np.array([0.5, 0.5, 0.5])  # meters; prevents gate from being too tight
    dx, dy, dz = k*np.array([sx, sy, sz]) + margin

    # clamp to sensible ranges so it doesn’t explode or collapse
    dx = float(np.clip(dx, 0.8, 6.0))
    dy = float(np.clip(dy, 0.5, 4.0))
    dz = float(np.clip(dz, 0.5, 3.0))

    return center0, (dx, dy, dz)

from sklearn.cluster import DBSCAN

def track_cart_auto_gate(xyz_files, uv0, roi_half=25, eps=0.35, min_samples=20):
    # Load first frame and estimate gate
    npz0 = np.load(xyz_files[0])
    xyz0 = npz0[npz0.files[0]]

    center0, (dx, dy, dz) = estimate_gate_from_first_frame(xyz0, uv0, roi_half=roi_half)
    centers = [center0]
    vel = np.zeros(3)

    for t in range(1, len(xyz_files)):
        npz = np.load(xyz_files[t])
        xyz = npz[npz.files[0]]

        pred = centers[-1] + vel

        pts = xyz.reshape(-1, 3)
        pts = pts[np.isfinite(pts).all(axis=1)]

        # 3D gate around predicted location
        m = (
            (np.abs(pts[:,0] - pred[0]) < dx) &
            (np.abs(pts[:,1] - pred[1]) < dy) &
            (np.abs(pts[:,2] - pred[2]) < dz)
        )
        cand = pts[m]

        if len(cand) < min_samples:
            centers.append(pred)
            continue

        # cluster to isolate cart
        labels = DBSCAN(eps=eps, min_samples=min_samples).fit_predict(cand)
        good = labels[labels != -1]
        if len(good) == 0:
            centers.append(pred)
            continue

        best = np.bincount(good).argmax()
        cluster = cand[labels == best]

        center = np.median(cluster, axis=0)
        centers.append(center)

        # update velocity (smooth)
        vel = 0.7 * vel + 0.3 * (centers[-1] - centers[-2])

    return np.array(centers), (dx, dy, dz)

xyz_files = [f"xyz/depth000{to_three_digits(i)}.npz" for i in range(299)]
uv0 = (720, 689)  # known golf cart pixel in frame 0
cart_centers_3d, gate = track_cart_auto_gate(xyz_files, uv0)
print("Auto gate dx,dy,dz:", gate)
```

```python
import numpy as np
xyz = np.load("xyz/depth000000.npz")
xyz = xyz[xyz.files[0]]

def nearby_pixels_within_percent(xyz, u0, v0, radius_px=5000, pct=0.30):
    """
    xyz: (H,W,3) array of [X,Y,Z] in meters for each pixel
    (u0,v0): seed pixel known to be on the golf cart
    radius_px: search window radius in pixels (square window)
    pct: percent threshold (0.20 = ±20%)
    
    Returns:
      mask_roi: (h,w) boolean mask over the ROI window
      coords_uv: Nx2 array of (u,v) pixel coords in full image
      pts_xyz:   Nx3 array of corresponding 3D points
      d0: seed distance
    """
    H, W, _ = xyz.shape
    u0 = int(np.clip(u0, 0, W-1))
    v0 = int(np.clip(v0, 0, H-1))

    p0 = xyz[v0, u0]
    if not np.isfinite(p0).all():
        raise ValueError("Seed pixel has NaN/Inf XYZ. Pick a different pixel or use a patch median.")

    d0 = float(np.linalg.norm(p0))
    lo, hi = (1 - pct) * d0, (1 + pct) * d0

    # ROI bounds
    u1 = max(0, u0 - radius_px)
    u2 = min(W-1, u0 + radius_px)
    v1 = max(0, v0 - radius_px)
    v2 = min(H-1, v0 + radius_px)

    roi = xyz[v1:v2+1, u1:u2+1]  # (h,w,3)
    valid = np.isfinite(roi).all(axis=2)

    dist = np.linalg.norm(roi, axis=2)  # (h,w)
    mask_roi = valid & (dist >= lo) & (dist <= hi)

    # Convert mask to full-image (u,v) coords
    ys, xs = np.where(mask_roi)          # indices inside ROI
    coords_uv = np.column_stack([xs + u1, ys + v1])  # (u,v) in full image
    pts_xyz = roi[mask_roi]

    return mask_roi, coords_uv, pts_xyz, d0
```

```python
import numpy as np
from PIL import Image

def nearby_pixels_with_overlay(xyz, u0, v0, rgb_path="/rgb/left000000.png",
                               radius_px=40, pct=0.20,
                               overlay_rgba=(255, 0, 0, 120)):
    """
    xyz: (H,W,3) array of [X,Y,Z] in meters
    (u0,v0): seed pixel on the golf cart
    rgb_path: path to corresponding RGB image
    radius_px: ROI half-size in pixels
    pct: threshold fraction (0.20 => ±20%)
    overlay_rgba: (R,G,B,A) for mask overlay (A=0..255)

    Returns:
      full_mask: (H,W) bool mask where True = within threshold
      overlay_img_rgba: (H,W,4) uint8 RGBA image (RGB with overlay)
      coords_uv: Nx2 (u,v) coordinates of selected pixels
      pts_xyz:   Nx3 xyz points of selected pixels
      d0: seed distance
    """
    H, W, _ = xyz.shape
    u0 = int(np.clip(u0, 0, W-1))
    v0 = int(np.clip(v0, 0, H-1))

    p0 = xyz[v0, u0]
    if not np.isfinite(p0).all():
        raise ValueError("Seed pixel has NaN/Inf XYZ. Pick a different pixel or use a patch median.")

    d0 = float(np.linalg.norm(p0))
    lo, hi = (1 - pct) * d0, (1 + pct) * d0

    # ROI bounds
    u1 = max(0, u0 - radius_px)
    u2 = min(W-1, u0 + radius_px)
    v1 = max(0, v0 - radius_px)
    v2 = min(H-1, v0 + radius_px)

    roi = xyz[v1:v2+1, u1:u2+1]  # (h,w,3)
    valid = np.isfinite(roi).all(axis=2)
    dist = np.linalg.norm(roi, axis=2)   # (h,w)

    mask_roi = valid & (dist >= lo) & (dist <= hi)

    # full-size mask
    full_mask = np.zeros((H, W), dtype=bool)
    full_mask[v1:v2+1, u1:u2+1] = mask_roi

    # coords and xyz points for selected pixels
    ys, xs = np.where(mask_roi)
    coords_uv = np.column_stack([xs + u1, ys + v1])  # (u,v)
    pts_xyz = roi[mask_roi]

    # --- build overlay image ---
    # Load RGB
    rgb_img = Image.open(rgb_path).convert("RGBA")
    if rgb_img.size != (W, H):
        raise ValueError(f"RGB image size {rgb_img.size} does not match xyz size {(W,H)}")

    # Create a transparent overlay layer
    overlay = Image.new("RGBA", (W, H), (0, 0, 0, 0))
    overlay_arr = np.array(overlay)

    # Apply overlay color where mask is true
    r, g, b, a = overlay_rgba
    overlay_arr[full_mask] = (r, g, b, a)

    # Mark the seed pixel clearly (yellow, full opacity)
    overlay_arr[v0, u0] = (255, 255, 0, 255)

    overlay = Image.fromarray(overlay_arr, mode="RGBA")

    # Composite overlay on top of RGB
    overlay_img = Image.alpha_composite(rgb_img, overlay)
    overlay_img_rgba = np.array(overlay_img)

    return full_mask, overlay_img_rgba, coords_uv, pts_xyz, d0
```

```python
import numpy as np

import numpy as np

def medoid_and_average_xyz(coords_uv, xyz):
    """
    coords_uv: Nx2 array of (u,v) pixel coords
    xyz: (H,W,3) array of [X,Y,Z] for each pixel

    Returns:
      (u_med, v_med): medoid pixel (closest actual pixel to centroid)
      avg_xyz: (X_mean, Y_mean, Z_mean) over all selected pixels (finite only)
    """
    coords_uv = np.asarray(coords_uv)
    if coords_uv.ndim != 2 or coords_uv.shape[1] != 2 or coords_uv.shape[0] == 0:
        raise ValueError("coords_uv must be Nx2 with at least 1 point.")

    u = coords_uv[:, 0].astype(float)
    v = coords_uv[:, 1].astype(float)

    # --- medoid pixel (closest to centroid in pixel space) ---
    cu, cv = u.mean(), v.mean()
    d2 = (u - cu) ** 2 + (v - cv) ** 2
    idx = int(np.argmin(d2))
    u_med = int(coords_uv[idx, 0])
    v_med = int(coords_uv[idx, 1])

    # --- average XYZ over all selected pixels ---
    pts = xyz[coords_uv[:, 1].astype(int), coords_uv[:, 0].astype(int)]  # (N,3)
    valid = np.isfinite(pts).all(axis=1)
    pts = pts[valid]
    if pts.shape[0] == 0:
        raise ValueError("All XYZ points for coords_uv are NaN/Inf.")

    avg_xyz = tuple(np.mean(pts, axis=0).tolist())  # (X_mean, Y_mean, Z_mean)

    return (u_med, v_med), avg_xyz


(u_med, v_med), avg_range = medoid_and_average_xyz(coords_uv, xyz=xyz)
print("Medoid pixel:", (u_med, v_med))
print("Average 3D distance (m):", avg_range)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

def animate_ego_and_cart(
    reversed_ego_xy,
    reversed_avg_xyz_world,
    points_per_sec=30,
    out_mp4="ego_cart.mp4",
    show=True
):
    ego = np.asarray(reversed_ego_xy)
    cart = np.asarray(reversed_avg_xyz_world)

    # align length
    N = min(len(ego), len(cart))
    ego = ego[:N]
    cart = cart[:N, :3]

    # finite masks
    ego_ok = np.isfinite(ego).all(axis=1)
    cart_ok = np.isfinite(cart[:, :2]).all(axis=1)

    # We'll still animate frame-by-frame; if a frame is invalid, we just don't add that point.
    # Swap axes for plotting: x=Y, y=X
    ego_xy_plot = np.column_stack([ego[:, 1], ego[:, 0]])
    cart_xy_plot = np.column_stack([cart[:, 1], cart[:, 0]])

    fig, ax = plt.subplots()
    ax.set_title(f"Ego + Golf Cart ({points_per_sec} frames/sec, 2 points/frame)")
    ax.set_xlabel("Y (left)")
    ax.set_ylabel("X (forward)")
    ax.grid(True)
    ax.axis("equal")

    # set bounds using all valid points from both
    all_pts = np.vstack([
        ego_xy_plot[ego_ok],
        cart_xy_plot[cart_ok]
    ])
    xmin, ymin = all_pts.min(axis=0)
    xmax, ymax = all_pts.max(axis=0)
    pad_x = 0.05 * (xmax - xmin + 1e-9)
    pad_y = 0.05 * (ymax - ymin + 1e-9)
    ax.set_xlim(-30, 30)
    ax.set_ylim(-10, 50)

    # two scatters (colors as requested earlier)
    ego_scat = ax.scatter([], [], s=20, color="green", label="Ego")
    cart_scat = ax.scatter([], [], s=20, color="orange", label="Golf cart")
    ax.legend(loc="best")

    ego_drawn = []
    cart_drawn = []

    def init():
        ego_scat.set_offsets(np.empty((0, 2)))
        cart_scat.set_offsets(np.empty((0, 2)))
        return ego_scat, cart_scat

    def update(i):
        # add one ego point (if valid)
        if ego_ok[i]:
            ego_drawn.append(ego_xy_plot[i])
            ego_scat.set_offsets(np.array(ego_drawn))

        # add one cart point (if valid)
        if cart_ok[i]:
            cart_drawn.append(cart_xy_plot[i])
            cart_scat.set_offsets(np.array(cart_drawn))

        return ego_scat, cart_scat

    interval_ms = 1000.0 / float(points_per_sec)
    anim = FuncAnimation(
        fig, update, frames=N, init_func=init,
        interval=interval_ms, blit=True
    )

    # Save MP4 (needs ffmpeg)
    try:
        from matplotlib.animation import FFMpegWriter
        anim.save(out_mp4, writer=FFMpegWriter(fps=points_per_sec, bitrate=1800))
        print(f"Saved: {out_mp4}")
    except Exception as e:
        print("MP4 save failed (likely missing ffmpeg). Error:", e)

    if show:
        plt.show()
    else:
        plt.close(fig)

# Usage:
animate_ego_and_cart(-ego_xy, -avg_xyz_world, points_per_sec=30, out_mp4="ego_cart.mp4", show=True)
```
